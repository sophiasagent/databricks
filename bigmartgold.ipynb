```python
spark.sql("""

CREATE OR REPLACE VIEW salesorder.bigmart.bigmart_sales_gold AS

SELECT *
FROM (

    SELECT *,
    
           ROW_NUMBER() OVER (
               PARTITION BY Item_Identifier
               ORDER BY ingestion_timestamp DESC
           ) AS rn

    FROM salesorder.bigmart.bigmart_sales

)

WHERE rn = 1

""")
```
